```python
import numpy as np 
import matplotlib.pyplot as plt

file_paths = [
    "./data_NRS06032024-2_FTIR/25-07-09_NRS06032024-2_TLW-2_BLW-4_S-4_H-3_sample_res-1_sample-500 scans_background-500 scans_range-7500-380_aperture-6mm_detector_MCT_Beamsplitter_KBr_vel-20khz_obj-36x_lens_cap.2.0.dpt"
]

FTIR_data = []

wn_lower_bound = 600
wn_upper_bound = 1100

for filepath in file_paths:
    data = np.loadtxt(filepath, delimiter=',')

    x_data = data[:,0]
    y_data = data[:,1]

    mask = (x_data >= wn_lower_bound) & (x_data <= wn_upper_bound)
    
    x_data_filtered = x_data[mask]
    y_data_filtered = y_data[mask]

    filtered_data = (x_data_filtered, y_data_filtered)

    FTIR_data.append(filtered_data)

plt.figure(figsize=(12,8))
for index, data in enumerate(FTIR_data, start=1):
    x_data = data[0]
    y_data = data[1]
    plt.plot(x_data, y_data, color="purple", linewidth=2)
    plt.grid(True, alpha=0.5)
    plt.xlabel("wavenumber", fontsize=12)
    plt.ylabel("Reflectance", fontsize=12)
    plt.title("test")

plt.show()
```
