```python
import numpy as np 
import matplotlib.pyplot as plt

def get_FTIR_data(file_paths, wn_bounds):

    FTIR_data = []

    for file in file_paths:

        data = np.loadtxt(file, delimiter='\t')

        wave_number = data[:,0]
        reflectance = data[:,1]

        mask = (wave_number >= wn_bounds[0]) & (wave_number <= wn_bounds[1])

        wave_number_filtered = wave_number[mask]
        reflectance_filtered = reflectance[mask]

        filtered_data = (wave_number_filtered, reflectance_filtered)
        FTIR_data.append(filtered_data)
    
    return FTIR_data


def plot_FTIR(FTIR_data, title, xlabel, ylabel, legend_keys):

    colors = colors = ['k', 'r', 'b', 'g', 'm']
    
    plt.figure(figsize=(15,10))
    for index, data in enumerate(FTIR_data):
        x = data[0]
        y = data[1]
        plt.plot(x, y, color=colors[index], linewidth=2, label=legend_keys[index])

    plt.grid(True, alpha=0.5)
    plt.xlabel(xlabel, fontsize=16)
    plt.ylabel(ylabel, fontsize=16)
    plt.title(title, fontsize=16)
    plt.tick_params(axis='both', which='major', labelsize=16)
    plt.legend(fontsize=14)
    plt.show()
```
