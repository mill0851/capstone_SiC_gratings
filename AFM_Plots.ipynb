```python
import numpy as np 
import matplotlib.pyplot as plt

file_paths = [
    "./data_NRS06032024-2_AFM/S-4_H-3_TLW-1_BLW-xx/NRS06032024_2-TLW-1-BLW-1-S-4-H-3-20X20um_250923_Z_Height_Forward_001_Line_154.txt",
    "./data_NRS06032024-2_AFM/S-4_H-3_TLW-1_BLW-xx/NRS06032024_2-TLW-1-BLW-2-S-4-H-3-20X20um_250923_Z_Height_Forward_002_Line_154.txt",
    "./data_NRS06032024-2_AFM/S-4_H-3_TLW-1_BLW-xx/NRS06032024_2-TLW-1-BLW-3-S-4-H-3-20X20um_250923_Z_Height_Forward_003_Line_154.txt",
    "./data_NRS06032024-2_AFM/S-4_H-3_TLW-1_BLW-xx/NRS06032024_2-TLW-1-BLW-4-S-4-H-3-20X20um_250923_Z_Height_Forward_005_Line_154.txt",
    "./data_NRS06032024-2_AFM/S-4_H-3_TLW-1_BLW-xx/NRS06032024_2-TLW-1-BLW-5-S-4-H-3-20X20um_250923_Z_Height_Forward_006_Line_156.txt"
]
afm_data = []

for filepath in file_paths:
    data = np.loadtxt(filepath, float, skiprows=4)

    with open (filepath, 'r') as file:
        lines = file.readlines()
        x_units = lines[0]
        y_units = lines[1]

    # Put data in (um)
    if y_units == "y(nm)\n":
        data[:,1] = data[:,1] * 10**-3
    elif x_units == "x(nm)\n":
        data[:,0] = data[:,0] * 10**-3

    # Normalize data
    data[:,1] = data[:,1] - np.min(data[:,1])

    afm_data.append(data)

plt.figure(figsize=(12,8))
for index, data in enumerate(afm_data, start=1):
    x_data = data[:, 0]
    y_data = data[:, 1]
    plt.subplot(2,3,index)
    plt.plot(x_data, y_data, color="purple", linewidth=2)
    plt.grid(True, alpha=0.5)
    plt.xlabel("Tip Horizontal Position (um)", fontsize=12)
    plt.ylabel("Tip Vertical Position (um)", fontsize=12)
    plt.title("AFM: TLW-1 S-4 H-3 " + f'BLW-{index}')

plt.tight_layout(pad=2.0)
plt.show()
```
