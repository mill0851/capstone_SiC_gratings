```python
import numpy as np 
import matplotlib.pyplot as plt

def get_AFM_data(file_paths):
    AFM_data = []

    for filepath in file_paths:
        data = np.loadtxt(filepath, float, skiprows=4)

        with open(filepath, 'r') as file:
            lines = file.readlines()
            x_units = lines[0]
            y_units = lines[1]

        # Put data in (um)
        if y_units == "y(nm)\n":
            data[:,1] = data[:,1] * 10**-3
        elif x_units == "x(nm)\n":
            data[:,0] = data[:,0] * 10**-3

        # Normalize data
        data[:,1] = data[:,1] - np.min(data[:,1])

        AFM_data.append(data)
    
    return AFM_data


def plot_AFM(AFM_data, title, xlabel, ylabel, legend_keys):
    colors = ['k', 'r', 'b', 'g', 'm']
    
    plt.figure(figsize=(15,10))
    for index, data in enumerate(AFM_data):
        x = data[:, 0]
        y = data[:, 1]
        plt.plot(x, y, color=colors[index], linewidth=2, label=legend_keys[index])

    plt.grid(True, alpha=0.5)
    plt.xlabel(xlabel, fontsize=30)
    plt.ylabel(ylabel, fontsize=30)
    plt.title(title, fontsize=30)
    plt.tick_params(axis='both', which='major', labelsize=30)
    plt.legend(fontsize=30)
    plt.show()
```
